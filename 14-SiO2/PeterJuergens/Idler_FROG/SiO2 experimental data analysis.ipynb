```python
time=np.loadtxt("time_vector.txt")
wavelengths=np.loadtxt("wavelengths.txt")
intensity_time=np.loadtxt("intensity_time.txt")
intensity_spectrum=np.loadtxt("intensity_spectrum.txt")
phase_time=np.loadtxt("phase_time.txt")
phase_spectrum=np.loadtxt("phase_spectrum.txt")
```

## Temporal evolution of pulse properties

```python
plt.figure()
plt.plot(time, intensity_time)
plt.xlabel("Time (fs)")
plt.ylabel("Intensity (arb.u.)")
plt.show()

plt.figure()
plt.plot(time, phase_time*180./np.pi)
plt.ylabel("Phase")
plt.show()
```

## Spectral evolution of the pulse

```python
plt.figure()
plt.plot(wavelengths, intensity_spectrum)
plt.xlabel("Wavelength (nm)")
plt.ylabel("Intensity (arb.u.)")
plt.show()

plt.figure()
plt.plot(wavelengths, phase_spectrum*180./np.pi)
plt.xlabel("Wavelength (nm)")
plt.ylabel("Phase")
plt.show()
```
