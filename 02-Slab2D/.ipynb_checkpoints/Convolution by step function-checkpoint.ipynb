```python
import numpy as np
import matplotlib as mp
import matplotlib.pyplot as plt
```

```python
c=3E8
wavelength=800e-9
omega=2.*np.pi*c/wavelength
tau=10E-15
numperiods=10. #duration of the quantity to average
periods = 1.   # duration of the window for averaging
E0 = 1E9

# modevalue = 'full'
modevalue = 'valid'
# modevalue = 'same'

## Defines the function to convolve
def f(t, wavelength, E0):
    return E0*np.sin(omega*t+0.*np.pi/3.) #*np.exp(-.5*(t/tau)**2)

## Defines the envelope of convolution
def g(t, wavelength,periods):
    return np.heaviside(t+periods*np.pi/omega,0.5) * (1-np.heaviside(t-periods*np.pi/omega,.5))
```

```python
tmin = -numperiods*np.pi/omega; tmax=numperiods*np.pi/omega
t=np.linspace(tmin,tmax,10000)
t_step_t = np.diff(t); t_step = t_step_t[t_step_t.size-1] #last value of time step

tmin_door = -periods * np.pi/omega; tmax_door = periods * np.pi/omega
t_door    = np.arange(tmin_door, tmax_door, t_step)

signal     =np.convolve(f(t,wavelength,E0), g(t     ,wavelength,periods), mode=modevalue)
signal_door=np.convolve(f(t,wavelength,E0), g(t_door,wavelength,periods), mode=modevalue)
```

```python
plt.figure()
ax1 = plt.subplot(211)
plt.xlabel('Time')
plt.plot(t,f(t,wavelength,E0),'r-')
ax12 = ax1.twinx()

plt.plot(t, g(t,wavelength,periods), 'b--')
plt.plot(t_door, g(t_door,wavelength,periods), 'b+-')

ax2 = plt.subplot(212)
plt.plot(signal, 'b-')
plt.plot(signal_door, 'r-')
plt.show()
```
