```python
import numpy as np
from numpy import heaviside
import matplotlib.pyplot as plt
```

```python
wavelength = 800e-9; c=3E8
omega    = 2.*np.pi*c/wavelength
tau_rise = 2.*np.pi/omega
print tau_rise
tmax = 5*2.*np.pi/omega
t = np.linspace(0, tmax, 300)
J_func = np.ones((t.shape))
damperB = np.sin(np.pi/2. * (t-tmax)/tau_rise)**2 * (np.heaviside(t-tmax+tau_rise,0.5)) * (1.-np.heaviside(t-tmax,0.5)) #useful for the [t2, tmax] interval. 
damperA = 1.- np.heaviside(t-(tmax-tau_rise),0.5)
J_new = np.multiply( J_func, damperA+damperB )
```

```python
plt.figure()
plt.xlabel('Time')
plt.ylabel('Current')
plt.plot(t, J_func)
plt.plot(t, J_new)
plt.show()
```
