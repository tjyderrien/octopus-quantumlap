## Create field of arrows

```python
ds = xr.Dataset(
    {
        "Angle": np.arctan2(
            xav.sel(x=slice(-10, 10), y=slice(-10, 10), component="y"),
            xav.sel(x=slice(-10, 10), y=slice(-10, 10), component="x"),
        ),
        "Magnitude": np.sqrt(
            xav.sel(x=slice(-10, 10), y=slice(-10, 10), component="x") ** 2
            + xav.sel(x=slice(-10, 10), y=slice(-10, 10), component="y") ** 2
        ),
    }
)  # polar coordinates required for plotting arrows
```

```python
def _vectorplot(val):
    plot = hv.VectorField(
        data=ds.sel(t=val, z=0, method="nearest"),
        kdims=["x", "y"],
        vdims=["Angle", "Magnitude"],
    )
    for dim in plot.kdims:  # keep the units
        dim.unit = xav[f"{dim}"].units
    return plot
```

```python
arrow_plot = (
    hv.DynamicMap(_vectorplot, kdims="t")
    .redim.values(t=xav.t.values)
    .opts(color="blue", width=500, height=400)
)  # Zoom-in for seeing the  arrows!
# arrow_plot.kdims[0].unit = xav.t.units
arrow_plot
```

```python
# Fxa = xrft.fft(xav, dim="t", real_dim="t")
# Fxa = xrft.fft(xav, dim="t", real_dim="t", true_amplitude=True, true_phase=True)
Fxa = xrft.fft(xav_damped, dim="t", real_dim="t", true_amplitude=True, true_phase=True)
# Fxa = xrft.fft(xav2, dim="t", real_dim='t', true_amplitude=True, true_phase=True)
# WARNING: omega space is not physical here. 
```

```python
# TimeStep = 0.25 #au
omega = Fxa.freq_t
print(omega)
```

```python
# Fxa.sel(component="x")
```

```python
FxaX=Fxa.sel(component='x')
FxaY=Fxa.sel(component='y')
FxaZ=Fxa.sel(component='z')
```

```python
Fxa_norm2 = FxaX.real ** 2 + FxaX.imag ** 2 + FxaY.real ** 2 + FxaY.imag ** 2 + FxaZ.real ** 2 + FxaZ.imag ** 2
```

```python
# Fxa_norm2 = Fxa.sel(component='x') *  + Fxa.sel(component='y')**2 + Fxa.sel(component='z')**2
```

```python
HHG = np.log10(np.clip(omega**2*Fxa_norm2, 1E-30, 1E15)) #Larmor formula in computational units (time is still a timestep !)
```

```python
from scipy.constants import c, pi, hbar, e
selected_freqs = np.linspace(0, 23, 24)
print(selected_freqs)
```

```python
from scipy.constants import c, pi, hbar, e

# Meaning of freq_t
Hartree_to_eV = 27.21138505
omega_eV = 2*pi*Fxa.freq_t * Hartree_to_eV #Fourier[f(a x)] = 1/|a| f_Fourier ( nu_x / a) (valid with which convention?)
# print(omega_eV)
# Harmonic order
wavelength_m = 2.237E-6
omega0 = 2*pi*c/wavelength_m * hbar / e
print("Laser (eV): ", omega0)
# Application to the plot
HHG = HHG.assign_coords(freq_t = omega_eV/omega0) #we may have a factor pi or c in the story. 
HHG = HHG.sel(z=0, method="pad")
HHG = HHG.sel(freq_t=selected_freqs, method="nearest")
```

# Cleaning up current data

```python
import glob
import matplotlib.pyplot as plt
```

```python
filenames=glob.glob("output_iter/td.*")
sorted_integers = sorted([int(item.split('td.')[1]) for item in filenames])
```

```python
np.array(sorted_integers)/5
```

```python
fourier_transform = np.fft.fft(sorted_integers)
frequencies = np.fft.fftfreq(len(sorted_integers))
magnitudes = np.abs(fourier_transform)

# Plotting the magnitudes of the Fourier transform
plt.figure(figsize=(10, 5))
plt.plot(frequencies, magnitudes) #, use_line_collection=True)
plt.xlabel('Frequency')
plt.ylabel('Magnitude')
plt.xlim((-0.05,0.05))
plt.title('Fourier Transform Magnitudes')
plt.savefig('fourier_transform_magnitudes.png')
```
