```python
from sympy import *
```

```python
t = Symbol('t')
tau = Symbol('tau')
omega = Symbol('omega')
```

```python
def convolve(f, g, x, lower_limit, upper_limit):
    y = Symbol('y')
    h = g.subs(x, x - y)
    return integrate(f * h, (y, lower_limit, upper_limit))
```

```python
convolve(sin(omega*t),Heaviside(t+pi/omega)*(1-Heaviside(t-pi/omega)), t, -oo, oo)
```
