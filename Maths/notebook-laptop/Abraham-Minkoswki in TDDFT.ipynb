## A-M in Keldysh theory vs TDDFT (no induced fields)

Results from TDDFT on Si is [located here](https://www.overleaf.com/read/wtvdqpfmnchs). In this reference, we demonstrate the role of the optical refractive index in matching TDDFT simulation results with the Keldysh theory. Matching only takes places if writing 

$ I_{\mathrm in} = n \times I_{\mathrm out} $. 

### External vs internal fields

To verify consistency, assuming that Keldysh theory is written for E being internal field (already affected by the optical index), $\gamma$ coefficient is defined by

$\gamma = \omega \left(\frac{m^* E_g}{e E_{\mathrm{in}}}\right)^{\frac{1}{2}}$

By introducing the definition $I_{\mathrm{in}} = \frac{1}{2} c \varepsilon_0 \left| E_{\mathrm{in}} \right|^2$ into $\gamma$, in two steps, one obtains 

$ \left| E_{\mathrm{in}} \right| = \sqrt{\frac{2 I_{\mathrm{in}}}{c \varepsilon_0}}$ and then

$\gamma = \omega \left(\frac{m^* E_g}{e }\right)^{\frac{1}{2}} \left( \frac{2 I_{\mathrm{in}}}{c \varepsilon_0} \right)^{-1/4}$. 

We can now express the $\gamma$ coefficient as function of $I_{\mathrm{out}}$ using $I_{\mathrm{in}}=n \times I_{\mathrm{out}}$ and obtain

$\gamma = \omega \left(\frac{m^* E_g}{e }\right)^{\frac{1}{2}} \left( \frac{2 n I_{\mathrm{out}} }{c \varepsilon_0} \right)^{-1/4} = \omega \left(\frac{m^* E_g}{e E_{\mathrm{out}}}\right)^{\frac{1}{2}} n^{-1/4}$. 

### Comparison with replacing directly $E_{\mathrm{in}} = \sqrt n E_{\mathrm{out}}$

$\gamma = \omega \left(\frac{m^* E_g}{e E_{\mathrm{in}}}\right)^{\frac{1}{2}}$

becomes 

$\gamma = \omega \left(\frac{m^* E_g}{e \sqrt n E_{\mathrm{out}}}\right)^{\frac{1}{2}} = \omega \left(\frac{m^* E_g}{e E_{\mathrm{out}}}\right)^{\frac{1}{2}} n^{-\frac{1}{4}}$


## A-M in energy absorption of Au vs TDDFT (with induced fields)

Results of energy absorption from TDDFT in Au is (maybe not yet in full) [given here](https://www.overleaf.com/read/hpmncddprswv). It occurs that energy absorption computed from TDDFT (with including the induced fields) leads to agreement with the experiments if writing 

$ I_{\mathrm in} = \frac{1}{2} c \varepsilon_0 \times (\vec{E}_{\mathrm out} / \eta )^2 $ 

where $\eta = \left| E_{\mathrm out} \right| / \left| E_{\mathrm in} \right| = \left | \mathrm{Re}(n)^2 \right| = \left| \mathrm{Re}(\varepsilon) \right| $, as shown by [Yabana2012] in the case of "longitudinal geometry", i.e., when $\vec A(t) = \vec A_{\mathrm out} (t) + \vec A_{\mathrm ind} (t)$, i.E., when accounting for the induced fields. 

For Au irradiated by 800 nm photons, $\mathrm{Re}(\varepsilon)\sim-24$ (see [RefractiveIndex.info](https://refractiveindex.info/?shelf=main&book=Au&page=Johnson)).

## Origin of the optical index in Kohn-Sham theory

From Eq. (26) of [Keldysh1965], momentum conservation writes $\vec{p}+\vec{A}$ based on Houston functions. This field is *internal*, i.e., it is already affected by the refractive index (absence of induced fields). 

## Verifying demonstration with Sympy

```python
from IPython.display import display, Latex
from sympy import *
init_printing()
```

```python
gamma, E, E_g, omega = symbols("gamma E E_g omega")
E_in, E_out = symbols("E_in E_out")
I_in, I_out = symbols("I_in I_out")
epsilon_0, c, m, e, n, wavelength = symbols("epsilon_0 c m e n wavelength")
```

```python
gamma = omega * sqrt((m * E_g)/(e * E_in))
gamma.expand()
gamma.subs("E_in", sqrt(n)*E_out).expand()
# gamma.subs(omega,2*pi*c/wavelength)
```

```python
# I_in=n * I_out
# E_in=sqrt(n)*E_out
# I_out=1/2 * c * epsilon_0 * E_out**2
eq1=Eq(I_in, 1/2 * c * epsilon_0 * E_in**2)
E_in=solve(I_in - 1/2 * c * epsilon_0 * E_in**2, E_in)[1]
```

```python
gamma.subs("E_in", E_in).subs("I_in", n * I_out).subs("I_out", 1/2 * c * epsilon_0 * E_out**2).expand()
```

We indeed find that replacing $E_{in}$ by $\sqrt{n} E_{out}$ is indeed equivalent to go through $I_{in}$ and $I_{out}$. We deduce that Gruzdev2014:Fig1 may indeed contain a mistake.
