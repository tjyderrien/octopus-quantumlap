```python
import numpy as np
import matplotlib as mp
import matplotlib.pyplot as plt
```

```python
plt.figure(figsize=(15,5))
x1 = np.linspace(1,3,15)
x2 = np.power(10,x1)
y2 = x2
ax1 = plt.subplot(111)
ax1.plot(x2, y2)
# plt.minorticks_on()
# ax1.xaxis.set_ticks_position('bottom')
# ax1.xaxis.set_minor_locator(mp.ticker.NullLocator())

ax1.set_xscale('log') #their position is finally not so critical
ax1.set_yscale('log') 

# ax1.tick_params(which='major', width=2.00, length=5)
# ax1.tick_params(which='minor', width=0.75, length=2.5)

# ax1.xaxis.grid(True, which="major")
# ax1.yaxis.grid(True, which="major")

# Preparing major ticks
locmaj = mp.ticker.LogLocator(base=10,numticks=24) 
ax1.xaxis.set_major_locator(locmaj)
ax1.yaxis.set_major_locator(locmaj)

# ax1.xaxis.set_major_locator(mp.ticker.LogLocator(base=10, numticks="auto"))
# ax1.yaxis.set_major_locator(mp.ticker.LogLocator(base=10, numticks="auto"))

# Preparing minor ticks
# ax1.xaxis.set_minor_locator(mp.ticker.LogLocator(base=10, numticks=1000))

subdiv=np.arange(0.,10.,1); print(subdiv)
locmin = mp.ticker.LogLocator(base=10.0,subs=subdiv,numticks=100)
ax1.xaxis.set_minor_locator(locmin)
ax1.xaxis.set_minor_formatter(mp.ticker.NullFormatter())

# ax1.xaxis.set_minor_locator(mp.ticker.LogLocator(base=10, numticks="auto")) #no effect ?!

# plt.tight_layout()

```
