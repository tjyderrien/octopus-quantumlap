# Context

Widera project of Angela Vella. Find what is happening in the SiO$_2$ needle upon 2 THz beam. Answer seems to be "not much."

```python
from scipy.constants import e, h, hbar, c, epsilon_0, pi
import numpy as np
from spp_extended_theory.Keldysh.libKeldysh import GenerateKeldyshDatabase
```

```python
photon_energy=2E12*hbar #2 THz [Vella et al, Science Adv. 2021]
print(photon_energy/e, "eV")

wavelength=2*pi*c*hbar/photon_energy
print(wavelength*1E3, "mm")
```

```python
spot_radius=10e-6
tau=3e-12

energy_range=np.linspace(0, 0.1E-3, 100) #J

intensity_range=energy_range/(pi*spot_radius**2)/tau
print(intensity_range*1E-4)
```

```python
Field_range=np.sqrt(2*intensity_range/c/epsilon_0)
print(Field_range*1E-9, "V/nm")
```

```python
wPI=GenerateKeldyshDatabase(9*e, 1, wavelength, Field_range, 1000)
print(wPI)
```

```python
N_exc=wPI*tau
print(N_exc)
```
