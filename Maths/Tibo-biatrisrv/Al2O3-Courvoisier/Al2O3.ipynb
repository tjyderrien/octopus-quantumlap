```python
import spglib as spg
import numpy as np
```

```python
numbers=[13, 13, 13, 13, 8, 8, 8, 8, 8, 8]
a_trig = 5.17796 #Angstroms for trigonal cell
a = 4.7606; c=12.994 #Angstroms for rhombohedral cell
a_prime=np.sqrt(a**2/3+c**2/9)
cos_alpha_prime=(2*c**2-3*a**2)/(2*(c**2+3*a**2))
alpha_prime=np.arccos(cos_alpha_prime)
print("Trigonal interatomic distance", a_prime, "angstroms")
print("Trigonal angle: ", alpha_prime*180/np.pi, "degrees")
```

# Primitive cell acc. to Duke university

```python
# Corrected using Duke University database of space group 167
lattice1_reduced=np.array([[0.5,-0.5/3**0.5,1./3.], 
         [0, 1/3**0.5,1./3.],
         [-0.5, -0.5/3**0.5, 1./3.]])

lattice2_reduced=np.array([[1*np.sin(alpha_prime/2), 
                            -1/np.sqrt(3)*np.sin(alpha_prime/2), 
                            1*np.sqrt((4*np.cos(alpha_prime/2)**2-1)/3)
                           ], 
                           [0, 
                            1*2/np.sqrt(3)*np.sin(alpha_prime/2),
                            1*np.sqrt((4*np.cos(alpha_prime/2)**2-1)/3)
                           ],
                           [-1*np.sin(alpha_prime/2), 
                            -1*np.sin(alpha_prime/2)/np.sqrt(3), 
                            1*np.sqrt((4*np.cos(alpha_prime/2)**2-1)/3)
                          ]])
```

```python
lattice1_xyz=np.multiply(lattice1_reduced, [a, a, c])
lattice2_xyz=np.multiply(lattice2_reduced, [a_prime, a_prime, a_prime])
print(lattice1_xyz)
print(lattice2_xyz)
```

```python
print("Volume lattice 1: ", np.dot(lattice1_xyz[0], np.cross(lattice1_xyz[1], lattice1_xyz[2])), "angstroms 3")
print("Volume lattice 2: ", np.dot(lattice2_xyz[0], np.cross(lattice2_xyz[1], lattice2_xyz[2])), "angstroms 3")
```
