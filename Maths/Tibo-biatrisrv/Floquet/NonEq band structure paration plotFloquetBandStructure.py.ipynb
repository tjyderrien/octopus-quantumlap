# Main idea of this work

We prepare here the band structure of several solids in non-equilibrium situation. 
This is a generalization of what was pioneered in the reference [1], where non-eq band structure was given at a single k-point for many fields.

In this report, we extend this computation by two ways: 
- we employ a **Floquet Hamiltonian** where field is introduced using length-gauge (E.d) that requires dipolar matrix elements d (see Ref. [2]); 
- we acquire the dipolar matrix elements by a first-principle method (see Ref. [2]).
- then, when enough data is available, we generalize it to the full band structure. 

Already existing studies employing this method are available in Ref. [3-4] and others.

# Results for Au (space group 225)

We start with data from Au (see crystal information at https://materialsproject.org/materials/mp-81/). 

Presently, we have two datasets available for Au (with two different k-point paths, for random reasons). 

```python
# DATASET1 (Au, PBE): DatabasePath=OctopusData+"/09-Au/Au-1x1x1-k-16x16x16-Mesh0.377945225092-Mesh0.377945225092-T0e0-Te0.025eV-PBE-ME/static/"
# DATASET2 (Au, LDA, new k-point path)
# DatabasePath=OctopusData+"/09-Au/Au-1x1x1-k-16x16x16-Mesh0.377945225092-Mesh0.377945225092-T0e0-Te0.025eV-PBE-ME/static/"
DatabasePath="/nfsfileshare/octopus.data/09-Au/Au-1x1x1-k-16x16x16-Mesh0.377945225092-Mesh0.377945225092-T0e0-Te0.025eV-PBE-ME"
%cd $DatabasePath
import octopus_slabs.Libs.plotFloquetBandstructure as floquet
```

We start with computing Stockman-like diagrams, where electron quasi-energy (i.e., laser-dressed) levels at the Gamma point of the band structure are represented as function of field intensity for a given wavelength. 

```python
floquet.SingleKpointManyFields()
```

Now we generalize this to the full band structure for a given field, and observe the effect on the full band structure, represented on a path. This path was chosen following symmetry rules, and is introduced directly into Octopus, prior to compute the dipolar matrix elements. Due to symmetries, the path is fully sufficient to characterize the band structure in the frame of Floquet formalism in the dipolar approximation. 

```python
floquet.MultipleKpointsOneField(1E9)
```

```python
floquet.MultipleKpointsOneField(1E10)
```

**Warning about precision**. An important point is to have a sufficient amount of k-point available in the path (upon preparation of the band structure with octopus). For example, having 100 points available between two k-points of the k-point-path is usually too small. This can lead to distrust in the result, although a careful inspection does demonstrate an identical result between two identical k-points (i.e., Floquet band structure at gamma points are equal - this can be double-checked by magnifying the results in terminal mode - out of ipython jupyter).
