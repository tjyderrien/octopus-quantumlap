# Matrix inversion for changing frame of dipolar matrix elements

```python
import numpy as np
```

```python
a0=5.431 #interatomic distance in Angstroms
AngPerBohr=0.52917721067
a0_au=a0/AngPerBohr
print(a0_au)
```

```python
A=np.array([[0.000003833391, 0, 0], 
           [0, 0.000001366599, 0], 
           [0,0,0.000002582836]])
#A=np.array([[0.000003833391, 0.000001366599, 0.000002582836]])
```

```python
P=a0_au*0.5*np.array([[0,1,1], 
              [1,0,1],
              [1,1,0]]) #this is expression of (a_1,a_2,a_3) in (x,y,z). In Gantmacher, it is T (see page 60). 
```

```python
Pinv=np.linalg.inv(P)
```

# Lets put $d_x,d_y,d_z$ in a vector

Old basis for d is assumed to be $(a_1,a_2,a_3)$.
New basis will be $(x,y,z)$. 

Taking the vector V expressed in (a1,a2,a3), we can re-express into (x,y,z) by writing

```python
V=np.array([[0.000003833391], [0.000001366599], [0.000002582836]]) #selected matrix elements
W=np.dot(P,V)
Wprime=P @ V
Wsecond=Pinv @ V
```

```python
print(W), print(Wprime), print(Wsecond)
```
