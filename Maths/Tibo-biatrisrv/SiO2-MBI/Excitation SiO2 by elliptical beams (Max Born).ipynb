```python
lattice=[[0.5,-0.5*3**0.5,0], 
         [0.5, 0.5*3**0.5,0],
         [0  , 0, 1]]
```

```python
positions=[[0.522715, 0.522715, 0.500000], #Si
           [0.477285, 0.000000, 0.166667], #Si
           [0.000000, 0.477285, 0.833333], #Si
           [0.584969, 0.839262, 0.370677], #O
           [0.160738, 0.745706, 0.037343], #O
           [0.254294, 0.415031, 0.704010], #O
           [0.839262, 0.584969, 0.629323], #O
           [0.745706, 0.160738, 0.962657], #O
           [0.415031, 0.254294, 0.295990]  #O
          ]
```

```python
numbers=[14, 14, 14, 8, 8, 8, 8, 8, 8]
```

```python
cell=(lattice, positions, numbers)
```

```python
symmetry = get_symmetry_dataset(cell, symprec=1e-5) #unnit is in Angstroms
print(symmetry['number'])
```

```python
refine_cell(cell)
```

```python
standardize_cell(cell)
```

```python
symmetry = get_symmetry_dataset(cell, symprec=1e-5)
print(symmetry)
```
